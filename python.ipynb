```python
import sounddevice as sd
import librosa
import numpy as np

SAMPLE_RATE = 44100
BLOCK_SIZE = 4096
PITCH_STEPS = 5

print("🎤 Real-time voice changer started")
print("Speak now...")
print("🛑 Stop the cell to finish")

def audio_callback(indata, outdata, frames, time, status):
    if status:
        print(status)

    chunk = indata[:, 0].astype(np.float32)

    try:
        changed = librosa.effects.pitch_shift(
            y=chunk,
            sr=SAMPLE_RATE,
            n_steps=PITCH_STEPS
        )

        if len(changed) >= frames:
            outdata[:, 0] = changed[:frames]
        else:
            outdata[:, 0] = np.pad(
                changed,
                (0, frames - len(changed))
            )

    except Exception:
        outdata[:, 0] = chunk

try:
    with sd.Stream(
        samplerate=SAMPLE_RATE,
        blocksize=BLOCK_SIZE,
        channels=1,
        dtype="float32",
        callback=audio_callback
    ):
        while True:
            sd.sleep(1000)

except KeyboardInterrupt:
    print("\n🛑 Voice changer stopped successfully.")

finally:
    sd.stop()
    print("🔊 Audio devices released.")
    
```

```python
import sounddevice as sd
import librosa
import numpy as np

# ==============================
# VOICE CHANGER SETTINGS
# ==============================

SAMPLE_RATE = 44100
BLOCK_SIZE = 4096

# ==============================
# SELECT VOICE EFFECT
# ==============================

print("\n🎙️ REAL-TIME VOICE CHANGER")
print("==========================")
print("1. Normal Voice")
print("2. High Voice")
print("3. Deep Voice")

choice = input("\nChoose an effect (1-3): ")

if choice == "1":
    PITCH_STEPS = 0
    effect_name = "Normal"

elif choice == "2":
    PITCH_STEPS = 5
    effect_name = "High"

elif choice == "3":
    PITCH_STEPS = -5
    effect_name = "Deep"

else:
    print("❌ Invalid choice. Using Normal Voice.")
    PITCH_STEPS = 0
    effect_name = "Normal"

print(f"\n✅ Selected Effect: {effect_name}")
print(f"🎵 Pitch Shift: {PITCH_STEPS} semitones")

# ==============================
# AUDIO PROCESSING
# ==============================

def audio_callback(indata, outdata, frames, time, status):

    if status:
        print(status)

    # Get microphone audio
    chunk = indata[:, 0].astype(np.float32)

    try:

        # Apply pitch shift
        if PITCH_STEPS != 0:

            changed = librosa.effects.pitch_shift(
                y=chunk,
                sr=SAMPLE_RATE,
                n_steps=PITCH_STEPS
            )

        else:

            changed = chunk

        # Match output size
        if len(changed) >= frames:

            outdata[:, 0] = changed[:frames]

        else:

            outdata[:, 0] = np.pad(
                changed,
                (0, frames - len(changed))
            )

    except Exception as e:

        print("Processing error:", e)

        outdata[:, 0] = chunk


# ==============================
# START REAL-TIME VOICE CHANGER
# ==============================

print("\n🎤 Starting voice changer...")
print("Speak into your microphone.")
print("🛑 Stop the cell when finished.\n")

try:

    with sd.Stream(
        samplerate=SAMPLE_RATE,
        blocksize=BLOCK_SIZE,
        channels=1,
        dtype="float32",
        callback=audio_callback
    ):

        while True:
            sd.sleep(1000)

except KeyboardInterrupt:

    print("\n🛑 Voice changer stopped.")

finally:

    sd.stop()

    print("🔊 Audio devices released.")
    print("✅ Program finished.")
```

```python
import numpy as np
from scipy import signal
import sounddevice as sd

# ==========================================
# ROBOT VOICE EFFECT
# ==========================================

def robot_voice(audio, sample_rate):
    """
    Creates a basic robotic voice effect.
    """

    # Create a carrier frequency
    carrier_frequency = 80

    time = np.arange(len(audio)) / sample_rate

    carrier = np.sin(
        2 * np.pi * carrier_frequency * time
    )

    # Ring modulation
    robot = audio * carrier

    # Normalize audio
    max_value = np.max(np.abs(robot))

    if max_value > 0:
        robot = robot / max_value

    return robot.astype(np.float32)


print("🤖 Creating robot voice...")

robot_audio = robot_voice(
    audio_mono,
    SAMPLE_RATE
)

print("✅ Robot voice created.")
print("Audio shape:", robot_audio.shape)
```

```python
print("🤖 Playing robot voice...")

sd.play(
    robot_audio,
    samplerate=SAMPLE_RATE
)

sd.wait()

print("✅ Robot voice playback completed.")
```
